```python
from datetime import date
from pymongo import MongoClient
from bson import json_util
from common_utils.logging import get_logger
from common_utils.ingester import read_cosmosdb_json, write_raw
import json

# ======================================
# Logger set up
# ======================================
logger = get_logger('cosmosdb_ingestion')

# ======================================
# Config Path
# ======================================
dbutils.widgets.text('path',"")
config_path=dbutils.widgets.get("path")
print(config_path)

# ======================================
# Read Config
# ======================================
with open(config_path,'r') as f:
    config=json.load(f)

# Configuration
config_source=config['source']
config_target=config['target']
run_date=date.today().isoformat()

logger.info('Reading data from the table %s',config_source['collection_name'])
logger.info('Reading data from the source MongoDB in the %s format.',config_target['file_format'])

# ======================================
# Reading from MongoDB Database
# ======================================
conn_str=dbutils.secrets.get(scope='retail-platform-dev',key=config_source['connection_string'])
df=read_cosmosdb_json(spark,conn_str, config_source['database_name'], config_source['collection_name'] )

logger.info('Sample data...')
df.display()
logger.info('In total %s rows are present in the table %s',df.count(),config_source['collection_name'])
target_path=f"{config_target['base_path']}{config_target['folder']}/load_date={run_date}"

# ======================================
# Data stored at our local machine
# ======================================
write_raw(df, target_path,config_target['file_format'], config_target['mode'], options = None)
logger.info('Data stored at location %s',target_path)
```
